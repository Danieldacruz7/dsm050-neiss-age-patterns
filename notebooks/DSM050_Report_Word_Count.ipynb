## 1  Introduction

### 1.1  Research background

Sport and recreation can involve injuries whose patterns differ across age and activity. Rui, Ashman and Akinseye (2019) found age-related differences in injured body regions among patients aged 5–24 using a different US emergency-department survey. Their restricted age range motivates a wider lifespan comparison, although their case definition prevents direct numerical comparison with this study. NEISS provides a common surveillance framework across ages.

This project treats that as a visualisation problem rather than a modelling one. The question is not what predicts injury, but what the age structure of sport- and recreation-associated injury looks like, and whether its clinical composition differs across age groups.

Core presentations peak at age 14. Head and face injuries account for 52% before age five but 13% at ages 45–54, while trunk and spine injuries become more prominent in older groups. Diagnosis patterns also change with age. Activities have distinct age profiles, and care escalation rises across supported adolescent and adult groups. Unreliable estimates limit conclusions at the oldest ages.

### 1.2  What the outcome is, and what it is not

The unit of observation is a sport- and recreation-product-associated emergency-department presentation: a patient treated at a sampled Emergency Department (ED) whose record carries a product code that this project's taxonomy classifies as sport or recreation. Two limits follow, and they constrain every claim made here.

Product association is not participation. A record coded to bicycles may be a child struck while riding, an adult commuting, or someone who tripped over a stationary bicycle. The product was involved. The person was not necessarily playing a sport.

There is no denominator. NEISS samples emergency departments, not the population, and carries no measure of how many people cycled, swam or played basketball. No rate, risk or incidence can be computed, and none is claimed. Every quantity in this report is a count of presentations or a share of them.

### 1.3  Research objectives and questions

The objectives are to describe age and sex patterns in presentations and how their clinical composition varies, and to assess how activity definitions and reliability screening affect those conclusions. Five research questions address these objectives.

## 2  Data source

NEISS is a stratified probability sample of US hospital emergency departments with at least six beds and a 24-hour ED, run by the Consumer Product Safety Commission since 1972. Trained coders extract eligible records and code demographic, injury, product and narrative fields (CPSC, 2020).

NEISS exists to support product regulation, not clinical epidemiology, and that shapes the data. Body part and diagnosis use coarse fixed lists. However, named sporting activities can be coded even without equipment involvement (CPSC, 2020, p. 29). Equipment-free sport injuries are not automatically absent. Coding practice also moves: six fields (Body_Part_2, Diagnosis_2, Product_3, Hispanic, Alcohol, Drug) were added on 1 January 2019. All six are excluded from the core analysis, because including them would let later years qualify on evidence that earlier years could not carry.

Each record carries Stratum (hospital size class), PSU (the sampled hospital) and Weight (inverse selection probability, ratio-adjusted to national ED visit totals). The primary sampling unit is the hospital, not the patient, so variance is approximated between hospitals within strata. The new sample first contributed national estimates in 2024, and CPSC backcast its historical series using an overlap comparison (Schroeder, 2026). This analysis does not backcast or assume independent annual samples.

NEISS public-use files are freely accessible and de-identified. Raw narratives are omitted from the shared analytical files to minimise disclosure. De-identification is not a guarantee against re-identification or an institutional ethics exemption. Race is not analysed, and its completeness and recording process would require separate evaluation.

The project uses 2016–2025. A recent window limits the historical mixture of activity participation, without eliminating it.

## 3  Data overview and pre-processing

All ten annual files were downloaded from CPSC, checksummed, and validated against the canonical 25-column schema. One finding breaks a naive read: the 2017 file is cp1252, not UTF-8, and raises UnicodeDecodeError on a plain pd.read_csv. The loader tries three encodings and records which one worked rather than silently dropping the year.

Sixteen short input fragments are rejected and logged before analysis. Seventy-eight otherwise eligible records have unknown age and are excluded from this age-focused cohort. No numeric zero weights occur among structurally valid rows.

NEISS codes 201–223 as 1–23 months, and 201 also covers newborns. Code 0 means unknown age, not infancy (CPSC, 2025, p. 10). The initial loader wrongly retained 78 such records, including 27 Core. The corrected loader preserves raw and decoded ages and accepts documented age codes only.

### 3.1  Product classification and cohort construction

NEISS defines 1,124 product codes and there is no sport flag. The analytical population is therefore an explicit taxonomy, built by reviewing every candidate code against its official CPSC label and recorded in product_classification.csv with a rationale and review status for each code. Of 173 candidates, 84 are Core, 123 Extended, 50 reviewed and excluded, and none left unreviewed.

Core covers codes classified as sport and exercise. This does not establish active participation. Extended adds broader recreation: playground equipment, cycling, water recreation, scooters, skateboards and trampolines. Core is the analytical population and Extended the sensitivity comparison. Two boundary decisions are worth stating. Cycling and playground equipment are Extended, not Core, because both are as much transport and general childhood activity as sport and either is large enough to dominate a lifespan curve alone. Fishing is included as sport. Bleachers cannot establish eligibility, although a separate qualifying sport code can.

Eligibility is Product_1 or Product_2. Product_3 is excluded because it is empty before 2019 and would give later years more chances to qualify. Activity attribution takes the first product qualifying for the relevant cohort, otherwise Product_2. An excluded first product cannot override a qualifying second product. One row is one presentation: unique case numbers equal cohort rows in every year, so no deduplication is required. The result is 659,637 Core and 978,847 Extended presentations. One presentation per row and one variable per column follows tidy-data principles (Wickham, 2014).

### 3.2  Derived groupings, and an error that was caught

The 30 body parts and 31 diagnoses are collapsed into interpretable groups, because a 30-row heatmap is unreadable and several official codes are near-duplicates. Both mappings are generated by joining on the official CPSC label text, not typed by hand, and every row carries a verification status.

An earlier hand-entered version had diagnosis codes 62 (INTERNAL INJURY) and 64 (STRAIN, SPRAIN) transposed, along with five other diagnosis and four body-part codes. The historical diagnosis output was therefore wrong. The notebook now asserts the mapping content at load time, so a repeat would fail the run rather than produce a plausible figure.

### 3.3  Survey weighting and reportability

Two rules apply throughout. Pooled burden estimates are average annual national estimates. Figure 1 shows individual-year totals. For display, an adapted conservative working screen is used: an estimate is reportable only if all three conditions hold: at least 20 unweighted cases, a weighted estimate of at least 1,200, and a finite upper-bound coefficient of variation (CV) of at most 0.33. Percentages also require their own CV bound to pass. Any failure suppresses the displayed value. These thresholds adapt CPSC guidance to average annual estimates. They are not an official multi-year certification (CPSC, n.d.b).

The annual with-replacement approximation retains the full observed hospital roster, including zero-contributing hospitals for each domain. For annual standard errors s_y, Cauchy–Schwarz gives SD(sum of annual estimates) ≤ sum(s_y). Dividing by the number of years bounds the average-total SE without assuming independent years. Pooled proportions use hospital residuals N − pD, retaining numerator–denominator covariance. Finite-population and ratio-adjustment corrections are omitted, so conservatism is relative to the approximation.

### 4.1  RQ1: The lifespan distribution

An unweighted age histogram would describe the sample. A weighted histogram would also be valid. Figure 2 plots the average annual national estimate at each single year of age. Single-year resolution is deliberate. Banding would smooth away the sharpness of the adolescent peak and the way the two cohort curves separate in childhood and converge later.

Presentations peak at age 14 and fall ninefold by age 40. The male curve peaks at 15 and the female curve at 14. These are presentation totals, not sex-specific participation rates.

Figure 2. Average annual weighted national estimates at each single year of age. Ages failing the conservative working screen are not drawn. Presentations, not rates: NEISS carries no participation denominator.

### 4.2  RQ2: Anatomy across the lifespan

Figure 2 answers how many. It says nothing about what kind, and the composition question is where the clinical interest sits. Figure 3 is a heatmap of composition rather than volume: each column is an age band and sums to 100% before suppression, so it asks where on the body an injury at that age is, not how many injuries there are. That normalisation is what makes childhood and old age comparable when the volumes differ by an order of magnitude. A single-hue sequential ramp is used because the encoded quantity is magnitude with no meaningful midpoint. A diverging or rainbow scale would invent a threshold that does not exist. Regions are ordered head to toe so that a downward drift reads as a downward drift.

Head and face account for 52% of presentations before age five and 13% by the mid-forties, while trunk and spine move the other way, from 6% at ages 5–9 to 32% at 75 and over. Wrist and hand peak in the 10–14 band at 23%. Injury mechanism was not analysed. Knee rises from 4% to 11% and then declines, so the lower-limb story is real but smaller than the trunk story.

Figure 3. Weighted body-region composition within age band, Core cohort. Each column sums to 100% before suppression. Cells failing the working screen are suppressed and shown as ‘-’.

### 4.3  RQ3: Diagnosis across the lifespan

A second heatmap could also support comparison, but trajectories emphasise changes across ordered age groups (Heer, Bostock and Ogievetsky, 2010). Figure 4 uses trajectories instead, because the question is about shape and crossing points. Five families are drawn and the rest pooled, since eight or more lines on one axis is a legend-reading exercise. A labelled legend identifies the lines and the remainder is dashed. Distinct marker shapes supplement colour so the lines remain distinguishable in greyscale.

Strain and sprain rise steeply through adolescence to 29% at 15–17, remain near that level through ages 25–34, then fall to under 10% in the oldest band. Fracture and internal injury move the other way after mid-life, internal injury more than tripling from its mid-life minimum. Laceration is the childhood diagnosis, at 21% before age five and 5% by 10–14. The pooled remainder is large and rises with age. It reaches 52.1% at 65–74. Its heterogeneity limits this grouping and should not be interpreted as a single diagnosis.

Figure 4. Weighted diagnosis-family composition within age band, Core cohort. All families remain in denominators, and 65 of 66 cells pass screening.

### 4.4  RQ4: Activity-specific age signatures

The aggregate curve in Figure 2 is a mixture. It could arise from every activity sharing one age profile, or from sharply different profiles that happen to sum to a smooth curve, and the aggregate cannot distinguish those. Figure 5 conditions on activity: each panel is P(age given activity), so a panel with 200,000 presentations and one with 8,000 are comparable in shape. Small multiples support comparison across many series using consistent scales (Wilke, 2019). Fifteen overlaid lines would obscure individual profiles. Panels are ordered alphabetically for lookup, and each keeps a faint copy of the whole-cohort curve behind it as a shared reference.

The activity profiles reveal variation concealed by the aggregate curve. The team sports are narrow and adolescent. Exercise and fitness, weight training and fishing have broader supported age profiles. Only 163 of 255 selected activity-age cells pass. Missing segments cannot be interpreted as no injuries.

Figure 5a. Activity-specific age composition, first nine panels. Weighted Core, failed working screens omitted.

Figure 5b. Age distribution within each activity, Core cohort, weighted. The 15 largest activities with at least 3,000 records are shown in five-year bands and an 80+ group. Full denominators and all working screens apply.

### 4.5  Advanced analysis: clustering activity phenotypes

Figure 5 shows that activities differ. It does not show whether those differences fall into a small number of recognisable types, which is difficult to assess from fifteen panels alone. Figure 6 applies hierarchical agglomerative clustering with Ward linkage to square-root composition profiles, three broad age groups and five broad body regions, with equal block weights, presented as a clustered heatmap with the dendrogram attached.

This is a structured descriptive device: no test or p-value is reported. Ward linkage is one of several defensible choices. The clusters describe presentations, not activities: two sports may group together because their participants are similar ages.

Ten activities pass every marginal-feature screen. Five are excluded. Ward and average linkage use the same Euclidean representation, with 77.8% pairwise agreement, so the four-group partition, chosen for readability rather than optimised, is linkage-sensitive (SciPy, n.d.). Full-window and through-2023 partitions agree on the nine common passing activities. This conditional stability does not validate a universal activity taxonomy. The merge heights show dissimilarity, not stability or confidence. Fishing is a singleton under both methods. This describes its presentation profile, not its eligibility as sport.

Figure 6. Hierarchical clustering (Ward linkage) on equally weighted square-root age and body-region proportions, Core cohort. Rows share the dendrogram's order, and the heatmap shows both feature blocks. Complete passing profiles only.

### 4.6  RQ5: Care outcome by disposition

NEISS carries no severity score. Disposition records care outcome: whether the patient was treated and released, or had care escalated: transferred, admitted, held for observation, or died in the ED. Two implementation points changed the answer. The classification is made on the disposition code, not the label text. An earlier version tested whether the label began with “TREATED”, which is true of TREATED/EXAMINED AND RELEASED but equally of TREATED AND ADMITTED/HOSPITALIZED and TREATED AND TRANSFERRED, and so counted admitted and transferred patients as released. And LEFT WITHOUT BEING SEEN and UNKNOWN are excluded from the denominator: they are not care outcomes.

Among passing groups, escalation runs from 2.0% at ages 10–14 to 13.7% at 55–64. At 0–4, 65–74 and 75+, the estimated-total CV fails although the proportion CV passes. Published released shares allow approximate reconstruction of withheld escalation shares. Masking limits display, not recoverability. Admission thresholds may differ by age, so care patterns cannot be interpreted solely as injury severity.

Figure 7. Escalation beyond treat-and-release by age band, Core cohort, weighted. Denominator includes codes 1, 2, 4, 5 and 8. 9,701 Core records are excluded. Three escalation age groups fail the working screen and are omitted.

## 5  Sensitivity analyses

Six sensitivity checks assess how analytical choices affect the displayed results.

|  | Check | Result |
| --- | --- | --- |
| A | Extended versus Core | Maximum diagnosis shift 17.81 pp, 65/66 cells compared |
| B | Product 1 only | Maximum diagnosis shift 0.58 pp, 65/66 cells compared |
| C | Exclude 2020 | Maximum diagnosis shift 0.59 pp, 65/66 cells compared |
| D | 2016–2023 versus 2016–2025 | Maximum diagnosis shift 1.76 pp, 65/66 cells compared |
| E | Unweighted versus weighted | Maximum activity-age shift 6.37 pp, 163/255 cells compared |
| F | Equal-year mean versus pooled | Maximum activity-age shift 0.58 pp, 121/255 cells compared |

The 2024 sample change affects cross-year comparison (Schroeder, 2026). A common multiplier cancels from a proportion. Changes in relative hospital weights do not. Check D examines excluding the refreshed years, while E compares weighting choices. Neither isolates reweighting causally. Check A produces the largest diagnosis shift: 17.81 percentage points in “All other diagnoses” at ages 0–4 (47.5% Core, 29.7% Extended, rounded). Contributions of individual added activities were not decomposed.

Equal-year means give each year equal influence. Pooled proportions give larger estimated denominators more influence. Check F compares only 121/255 cells, limiting claims of small differences.

## 6  Critical engagement with AI

AI assistance was used for code drafting, dataset auditing and prose editing. The assistance history is preserved in AI_USAGE.md. ChatGPT subsequently rebuilt and executed the corrected analysis with permission. Assistant checks are distinguished from personal verification. Two examples were errors that reached a figure, and they were the same error twice.

The body-region and diagnosis-family mappings were originally produced by typing NEISS code numbers rather than joining to the official labels, and diagnosis codes 62 and 64 were transposed. The resulting diagnosis distribution was wrong despite plausible-looking code. The historical log records the error and correction. The current helper verifies code-label joins.

The same failure mode then recurred in section 4.6, classifying disposition by matching a text prefix instead of the code. Both illustrate why a plausible label match is insufficient. Current assertions check mappings and attribution. Analyst-defined groupings remain reviewable CSVs. These checks catch specified failures, not every possible error.

The general lesson is that AI output which is plausible in form can be wrong in content. Both source verification and numerical checks are needed. Neither code review nor output plausibility alone establishes correctness.

## 7  Conclusions and evaluation

### 7.1  What the figures show

RQ1: Core presentations peak at 14 and fall approximately ninefold by 40, and male and female peaks occur at 15 and 14 respectively.

RQ2: Head and face account for 52% before age five and 13% at 45–54, and trunk and spine rise from 6% at 5–9 to 32% at 75+.

RQ3: Strain and sprain become less prominent after mid-life, while fracture and internal injury become more prominent.

RQ4: Activities have distinct age profiles. Their mixture shapes the overall curve.

RQ5: Escalation rises across supported adolescent and adult groups. Oldest-group interpretation remains conditional on the screening rule.

### 7.2  What the figures cannot show

No rate, risk or incidence appears anywhere, because NEISS has no participation denominator. The declining volume after adolescence is equally consistent with fewer injuries per participant, fewer participants, or older people presenting to primary care instead, and this data cannot separate them. Product association is not participation. Disposition depends on admission practice as well as injury characteristics. It is neither a severity score nor a quantified bound on severity differences.

### 7.3  Evaluation of the visualisation choices

Normalising within age band, in Figures 3 and 4, is what makes childhood and old age comparable when the volumes differ by an order of magnitude. It also lets a thin age band look important, so every such figure carries the raw case count through the declared working screen and suppresses what fails it.

Trajectories rather than a second heatmap for RQ3 keeps the two figures doing different work, while a second heatmap would remain a defensible alternative. Small multiples for RQ4 was not a close call at fifteen series, and alphabetical ordering helps lookup but does not reveal an age ordering. Shared scales support comparison while suppression limits coverage.

Repeating the project, the effort spent on Figure 6, which is a structured description rather than a result, would be better spent on Location, which is populated throughout and was not used. Further improvements include official survey-variance implementation and alternative clinical groupings. Only 396/2,310 joint activity-age-anatomy cells pass: adequate marginal profiles do not establish adequate joint coverage.
